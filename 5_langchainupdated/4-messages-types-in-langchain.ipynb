Text Prompts with model.invoke() — Theory Notes
1. What is a Text Prompt?

A text prompt is a natural-language instruction or question given to a Large Language Model (LLM) to generate a response.

For example:

"Explain artificial intelligence in simple terms."

The prompt acts as the input to the model.
3. How it works

When we pass a text prompt to model.invoke():

The user creates a prompt.
The prompt is passed to model.invoke().
LangChain sends the input to the configured LLM.
The LLM processes the prompt.
The model generates a response.
LangChain returns the response as a message object for chat models.
4. Input and Output

Input:

A string containing the text prompt.

Output:

For a chat model, the output is generally an AIMessage object containing the generated response.

The actual generated text can be accessed using:

response.content
Text Prompt vs Chat Messages

A simple text prompt can be passed directly:

model.invoke("Explain machine learning")

# Types of Chat Messages in LangChain

In LangChain, chat models work with **structured messages**. Each message represents a different role in a conversation.

There are **three main types** of chat messages:

### 1. `SystemMessage`

A **SystemMessage** provides instructions, rules, personality, or behavior that the AI should follow.

**Purpose:** Defines **how the AI should behave**.

Example:

```python
SystemMessage(content="You are a helpful Python teacher.")
```

Think of it as:

> **Instructions given to the AI.**

---

### 2. `HumanMessage`

A **HumanMessage** represents the message or question provided by the user.

**Purpose:** Contains the **user's input/query**.

Example:

```python
HumanMessage(content="Explain Python functions.")
```

Think of it as:

> **What the user is asking the AI.**

---

### 3. `AIMessage`

An **AIMessage** represents the response generated by the AI model.

**Purpose:** Stores the **AI's response**.

Example:

```python
AIMessage(content="A Python function is a reusable block of code...")
```

Think of it as:

> **What the AI answered.**

---

## Conversation Flow

```text
SystemMessage
      ↓
"You are a helpful teacher."
      ↓
HumanMessage
      ↓
"Explain machine learning."
      ↓
AIMessage
      ↓
"Machine learning is..."
```

### Simple Example

```python
from langchain_core.messages import (
    SystemMessage,
    HumanMessage,
    AIMessage
)

messages = [
    SystemMessage(content="You are a helpful teacher."),
    HumanMessage(content="What is Python?"),
    AIMessage(content="Python is a programming language."),
    HumanMessage(content="What is it used for?")
]
```

The model uses these messages to understand the **role and context** of each part of the conversation.

### Quick Revision

| Message           | Represents                    | Main Purpose            |
| ----------------- | ----------------------------- | ----------------------- |
| **SystemMessage** | System/Developer instructions | Controls AI behavior    |
| **HumanMessage**  | User                          | Provides query/input    |
| **AIMessage**     | AI                            | Provides model response |

**Remember:**
**System → instructions**
**Human → question/input**
**AI → response**


## `ToolMessage` in LangChain

A **`ToolMessage`** is a type of chat message used to represent the **result returned by a tool after the AI calls that tool**.

It is mainly used in **tool calling / function calling** workflows.

### Basic Flow

```text
HumanMessage
     ↓
   AIMessage
     ↓
AI decides to call a tool
     ↓
    Tool
     ↓
ToolMessage
     ↓
   AIMessage
     ↓
Final answer
```

### What is its purpose?

`ToolMessage` sends the **tool's output back to the model** so that the model can use that information to generate the final response.

For example, suppose the AI calls a weather tool:

```text
HumanMessage:
"What is the weather in Mumbai?"

        ↓

AIMessage:
"Call weather_tool(city='Mumbai')"

        ↓

Tool executes

        ↓

ToolMessage:
"Temperature: 30°C, Condition: Sunny"

        ↓

AIMessage:
"The weather in Mumbai is 30°C and sunny."
```

### Important Concept

`ToolMessage` is **not the user's message** and it is **not the AI's final response**.

It represents:

> **The result produced by an external tool and returned to the AI.**

### Typical Structure

```python
from langchain_core.messages import ToolMessage

message = ToolMessage(
    content="Temperature is 30°C",
    tool_call_id="abc123"
)
```

The `tool_call_id` connects the tool result to the corresponding tool call made by the AI.

### Four Important Message Types

| Message         | Role                       |
| --------------- | -------------------------- |
| `SystemMessage` | Gives instructions to AI   |
| `HumanMessage`  | User's input               |
| `AIMessage`     | AI's response/tool request |
| `ToolMessage`   | Result returned by a tool  |

### Easy Way to Remember

```text
SystemMessage → "How should you behave?"
HumanMessage  → "What does the user want?"
AIMessage     → "What does the AI say/do?"
ToolMessage   → "What did the tool return?"
```

**Key point:** In an agentic LangChain application, `ToolMessage` is what allows the model to receive the **actual result of a tool execution** and continue reasoning from that result.


````python
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_core.messages import (
    SystemMessage,
    HumanMessage,
    AIMessage,
    ToolMessage
)

load_dotenv()

# Groq model
model = ChatGroq(
    model="llama-3.1-8b-instant",
    temperature=0
)

# All message types in a single list
messages = [

    # 1. System Message
    SystemMessage(
        content="You are a helpful AI assistant. Answer clearly and concisely."
    ),

    # 2. Human Message
    HumanMessage(
        content="What is Python?"
    ),

    # 3. AI Message
    AIMessage(
        content="Python is a high-level, general-purpose programming language."
    ),

    # 4. Human Message
    HumanMessage(
        content="What is Python mainly used for?"
    ),

    # 5. Tool Message
    ToolMessage(
        content="Python is commonly used for web development, AI, machine learning, automation, and data analysis.",
        tool_call_id="demo_tool_call_123"
    )
]

# Send all messages to the model
response = model.invoke(messages)

print(response.content)
````

````python
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_core.messages import (
    SystemMessage,
    HumanMessage,
    ToolMessage
)
from langchain_core.tools import tool

load_dotenv()

# --------------------------------------------------
# 1. Create Groq model
# --------------------------------------------------

model = ChatGroq(
    model="llama-3.1-8b-instant",
    temperature=0
)


# --------------------------------------------------
# 2. Create a custom tool
# --------------------------------------------------

@tool
def convert_usd_to_inr(usd: float) -> str:
    """Convert USD to INR using a fixed demo exchange rate."""
    
    rate = 87.50
    inr = usd * rate

    return f"${usd:.2f} USD = ₹{inr:.2f} INR"


# --------------------------------------------------
# 3. Bind tool to model
# --------------------------------------------------

model_with_tools = model.bind_tools(
    [convert_usd_to_inr]
)


# --------------------------------------------------
# 4. System + Human Message
# --------------------------------------------------

messages = [
    SystemMessage(
        content="""
        You are a helpful financial assistant.

        Rules:
        1. Be concise and accurate.
        2. If the user asks for USD to INR conversion,
           use the convert_usd_to_inr tool.
        3. Explain the final result clearly.
        """
    ),

    HumanMessage(
        content="Convert 250 USD into Indian Rupees."
    )
]


# --------------------------------------------------
# 5. First model invocation
# --------------------------------------------------

ai_response = model_with_tools.invoke(messages)

print("\n========== AI MESSAGE ==========")
print(ai_response)


# --------------------------------------------------
# 6. Check whether AI requested a tool
# --------------------------------------------------

print("\n========== TOOL CALLS ==========")
print(ai_response.tool_calls)


# Add AI response to conversation
messages.append(ai_response)


# --------------------------------------------------
# 7. Execute the requested tool
# --------------------------------------------------

for tool_call in ai_response.tool_calls:

    if tool_call["name"] == "convert_usd_to_inr":

        result = convert_usd_to_inr.invoke(
            tool_call["args"]
        )

        print("\n========== TOOL RESULT ==========")
        print(result)

        # Create ToolMessage
        tool_message = ToolMessage(
            content=result,
            tool_call_id=tool_call["id"]
        )

        messages.append(tool_message)


# --------------------------------------------------
# 8. Send ToolMessage back to Groq
# --------------------------------------------------

final_response = model_with_tools.invoke(messages)


# --------------------------------------------------
# 9. Final AI response
# --------------------------------------------------

print("\n========== FINAL AI RESPONSE ==========")
print(final_response.content)


# --------------------------------------------------
# 10. Usage Metadata
# --------------------------------------------------

print("\n========== USAGE METADATA ==========")
print(final_response.usage_metadata)


# --------------------------------------------------
# 11. Response Metadata
# --------------------------------------------------

print("\n========== RESPONSE METADATA ==========")
print(final_response.response_metadata)
````
