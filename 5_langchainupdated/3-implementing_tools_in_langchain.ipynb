```python
from langchain.tools import tool
@tool
def get_weather(location:str)->str:
    """get the weather of the city"""
    print (f"The weather of {location} is Sunny.")


from langchain.chat_models import init_chat_model

model = init_chat_model(
    "llama-3.3-70b-versatile",
    model_provider="groq"
)

tool_with_model=model.bind_tools([get_weather])
```

```python
response=tool_with_model.invoke("what is the weather in boston?")
for tool_call in response.tool_calls:
    print(f"Tool:{tool_call["name"]}")
    print(f"Args:{tool_call["args"]}")
```

```python
messages=[{"role":"user","content":"what is the weather in boston?"}]
ai_message=tool_with_model.invoke(messages)
messages.append(ai_message)
```

```python
for tool_call in ai_message.tool_calls:
    tool_result=get_weather.invoke(tool_call)
    messages.append(tool_result)
```

```python
final_response=tool_with_model.invoke(messages)
print(final_response.text)
```
