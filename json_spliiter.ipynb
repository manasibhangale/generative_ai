```python
import requests
from langchain_text_splitters import RecursiveJsonSplitter

# Fetch JSON from LangSmith API
url = "https://api.smith.langchain.com/openapi.json"

json_data = requests.get(url).json()

# Create JSON splitter
splitter = RecursiveJsonSplitter(
    max_chunk_size=300
)

# Split JSON into chunks
json_chunks = splitter.split_json(
    json_data=json_data
)

# Print chunks
for chunk in json_chunks[:5]:
    print(chunk)
    print("----------------")
```
