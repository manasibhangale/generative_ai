```python
from langchain_community.embeddings import OllamaEmbeddings
embeddings=OllamaEmbeddings(model="nomic-embed-text") #by default uses llama 2
```

```python
embeddings
```

```python
r1 = embeddings.embed_documents([
    "alpha is the first letter of the Greek alphabet",
    "gamma is the third letter of the Greek alphabet",
])

print(len(r1))      # Number of documents
print(len(r1[0]))   # Embedding dimension
```

```python
r1
```

```python
r2=embeddings.embed_query("which is the first letter of greek alphabet")
```

```python
r2
```

Recommended embedding models for Ollama

If you've pulled one of these models, use it instead:

from langchain_community.embeddings import OllamaEmbeddings

embeddings = OllamaEmbeddings(model="nomic-embed-text")

or

embeddings = OllamaEmbeddings(model="mxbai-embed-large")

First, pull the embedding model if you haven't already:

ollama pull nomic-embed-text

or

ollama pull mxbai-embed-large

Then you can generate embeddings:

r1 = embeddings.embed_documents([
    "alpha is the first letter of the Greek alphabet",
    "gamma is the third letter of the Greek alphabet",
])

print(len(r1))      # Number of documents
