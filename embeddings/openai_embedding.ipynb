```python
from langchain_openai import OpenAIEmbeddings
embedings_1024=OpenAIEmbeddings(model="text-embedding-3-large",dimensions=1024)
```

```python
text="this is the tutorial on openai embeddings"
query_result1=embedings_1024.embed_query("text")
query_result1
```

```python
len(query_result1)
```

```python
from langchain_community.document_loaders import TextLoader
loader=TextLoader("speech.txt")
loader
```

```python
text_documents=loader.load()
text_documents
```

```python
#vector embeddding and vectore storeDB
from langchain_community.vectorstores import Chroma
db=Chroma.from_documents(text_documents,embedings_1024)
retrieved_results=db.similarity_search("Dr. Kalam took up academic pursuit as Professor, Technology & Societal Transformation at Anna University, Chennai from November 2001 and was involved in teaching and research tasks.")
print(retrieved_results)
```
