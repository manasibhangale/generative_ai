```python
from langchain_community.document_loaders import WebBaseLoader
from langchain_text_splitters import HTMLHeaderTextSplitter

# URL
url = "https://www.geeksforgeeks.org/machine-learning/machine-learning/"

# Load HTML from URL
loader = WebBaseLoader(url)
docs = loader.load()

# Extract HTML content
html_string = docs[0].page_content

# Define headers to split on
headers_to_split_on = [
    ("h1", "Header 1"),
    ("h2", "Header 2"),
    ("h3", "Header 3"),
]

# Create splitter
header_splitter = HTMLHeaderTextSplitter(
    headers_to_split_on=headers_to_split_on
)

# Split HTML
html_header_splitter = header_splitter.split_text(html_string)

# Print output
for doc in html_header_splitter:
    print(doc)
    print("-------------------")
```
